```python
import pandas as pd

# Load the telecom customer churn dataset
df = pd.read_csv("data/Telco-Customer-Churn.csv")

# Display the first five rows
df.head()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
print("Column Names:")
print(df.columns.tolist())

print("\nDataset Shape:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nDataset Information:")
df.info()

print("\nFirst Five Records:")
display(df.head())

print("\nLast Five Records:")
display(df.tail())

print("\nColumn Data Types:")
print(df.dtypes.value_counts())
```

```python
# STEP 8: Analyze missing values and customer churn distribution

# Check missing values in each column
print("Missing Values in Each Column:")
print(df.isnull().sum())

# Calculate total missing values
print("\nTotal Missing Values:", df.isnull().sum().sum())

# Check duplicate records
print("\nDuplicate Records:", df.duplicated().sum())

# Count customers who stayed and customers who left
print("\nCustomer Churn Distribution:")
print(df["Churn"].value_counts())

# Calculate the percentage of each churn category
print("\nCustomer Churn Percentage:")
print((df["Churn"].value_counts(normalize=True) * 100).round(2))

# Visualize the churn distribution
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(data=df, x="Churn")
plt.title("Customer Churn Distribution")
plt.xlabel("Churn Status")
plt.ylabel("Number of Customers")
plt.show()
```

```python
# STEP 9: Clean and preprocess the customer churn dataset

# Create a copy to preserve the original dataset
df_clean = df.copy()

# Remove extra spaces from column names and text values
df_clean.columns = df_clean.columns.str.strip()

for col in df_clean.select_dtypes(include="object").columns:
    df_clean[col] = df_clean[col].str.strip()

# Convert TotalCharges to numeric values
df_clean["TotalCharges"] = pd.to_numeric(
    df_clean["TotalCharges"], errors="coerce"
)

# Check missing values before handling them
print("Missing values before cleaning:")
print(df_clean.isnull().sum())

# Remove duplicate records if present
duplicates = df_clean.duplicated().sum()
print("\nDuplicate records found:", duplicates)

df_clean = df_clean.drop_duplicates()

# Fill missing TotalCharges using the median
median_total_charges = df_clean["TotalCharges"].median()

df_clean["TotalCharges"] = df_clean["TotalCharges"].fillna(
    median_total_charges
)

# Check missing values after cleaning
print("\nMissing values after cleaning:")
print(df_clean.isnull().sum())

print("\nCleaned dataset shape:", df_clean.shape)

# Display the first five cleaned records
display(df_clean.head())
```

```python
# STEP 10: Calculate descriptive statistics for numerical features

# Select numerical columns from the cleaned dataset
numeric_data = df_clean.select_dtypes(include=["number"])

# Calculate mean, median, mode, variance, and standard deviation
statistics = pd.DataFrame({
    "Mean": numeric_data.mean(),
    "Median": numeric_data.median(),
    "Mode": numeric_data.mode().iloc[0],
    "Variance": numeric_data.var(),
    "Standard Deviation": numeric_data.std(),
    "Minimum": numeric_data.min(),
    "Maximum": numeric_data.max()
})

# Display the statistical summary
print("Descriptive Statistics of Numerical Features:")
display(statistics.round(2))
```

```python
# STEP 11: Visualize numerical features and detect potential outliers

import matplotlib.pyplot as plt
import seaborn as sns

# Select numerical columns, excluding binary identifier-like features
numeric_cols = [
    col for col in df_clean.select_dtypes(include="number").columns
    if df_clean[col].nunique() > 2
]

# 1. Plot histograms to understand data distributions
df_clean[numeric_cols].hist(
    figsize=(12, 8),
    bins=25,
    edgecolor="black"
)

plt.suptitle("Histograms of Numerical Features")
plt.tight_layout()
plt.show()

# 2. Plot box plots to visually inspect outliers
for col in numeric_cols:
    plt.figure(figsize=(8, 3))
    sns.boxplot(x=df_clean[col])
    plt.title(f"Box Plot of {col}")
    plt.tight_layout()
    plt.show()

# 3. Detect potential outliers using the IQR method
outlier_results = []

for col in numeric_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[col] < lower_bound) |
        (df_clean[col] > upper_bound)
    ]

    outlier_results.append({
        "Feature": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers)
    })

# Display the outlier summary
outlier_summary = pd.DataFrame(outlier_results)
print("Potential Outliers Detected Using IQR:")
display(outlier_summary.round(2))
```

```python
# STEP 12: Visualize categorical features and analyze correlations

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 1. Plot customer counts for important categorical features
categorical_cols = [
    "Contract",
    "PaymentMethod",
    "InternetService",
    "TechSupport",
    "PaperlessBilling"
]

for col in categorical_cols:
    plt.figure(figsize=(9, 4))
    sns.countplot(data=df_clean, x=col)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Number of Customers")
    plt.xticks(rotation=25, ha="right")
    plt.tight_layout()
    plt.show()

# 2. Calculate churn rate for each category
for col in categorical_cols:
    churn_rate = (
        df_clean.groupby(col, observed=False)["Churn"]
        .apply(lambda x: (x == "Yes").mean() * 100)
        .sort_values(ascending=False)
    )

    plt.figure(figsize=(9, 4))
    sns.barplot(x=churn_rate.index, y=churn_rate.values)
    plt.title(f"Churn Rate by {col}")
    plt.xlabel(col)
    plt.ylabel("Churn Rate (%)")
    plt.xticks(rotation=25, ha="right")
    plt.tight_layout()
    plt.show()

# 3. Create a correlation heatmap for numerical features
numeric_data = df_clean.select_dtypes(include="number")

plt.figure(figsize=(8, 5))
sns.heatmap(
    numeric_data.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap of Numerical Features")
plt.tight_layout()
plt.show()
```

```python
# STEP 13: Separate features and target, then split into training and testing sets

from sklearn.model_selection import train_test_split

# Remove customer ID because it is only an identifier
df_model = df_clean.drop(columns=["customerID"], errors="ignore").copy()

# Separate input features (X) and target variable (y)
X = df_model.drop(columns=["Churn"])
y = df_model["Churn"].map({"No": 0, "Yes": 1})

# Display feature and target information
print("Number of input features:", X.shape[1])
print("Number of records:", X.shape[0])

print("\nTarget distribution:")
print(y.value_counts().rename(index={0: "No Churn", 1: "Churn"}))

# Split data into 80% training and 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True).round(3))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True).round(3))
```

```python
# STEP 14: Preprocess features and train Logistic Regression

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression

# Identify numerical and categorical input features
numeric_features = X_train.select_dtypes(include=["number"]).columns.tolist()
categorical_features = X_train.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

# Preprocess numerical features
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Preprocess categorical features
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both preprocessing methods
preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

# Create the complete machine-learning pipeline
logistic_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(max_iter=2000, random_state=42))
])

# Train the model using training data
logistic_model.fit(X_train, y_train)

print("Logistic Regression model trained successfully!")
print("Numerical features:", len(numeric_features))
print("Categorical features:", len(categorical_features))
```

```python
# STEP 15: Evaluate Logistic Regression using classification metrics

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    RocCurveDisplay
)
import matplotlib.pyplot as plt

# Predict churn classes and churn probabilities on test data
y_pred = logistic_model.predict(X_test)
y_prob = logistic_model.predict_proba(X_test)[:, 1]

# Calculate performance metrics
metrics = {
    "Accuracy": accuracy_score(y_test, y_pred),
    "Precision": precision_score(y_test, y_pred, zero_division=0),
    "Recall": recall_score(y_test, y_pred, zero_division=0),
    "F1-Score": f1_score(y_test, y_pred, zero_division=0),
    "ROC-AUC": roc_auc_score(y_test, y_prob)
}

# Display metric results
print("Model Performance:")
for metric, score in metrics.items():
    print(f"{metric}: {score:.4f}")

# Display detailed classification report
print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["No Churn", "Churn"],
    zero_division=0
))

# Display confusion matrix
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    display_labels=["No Churn", "Churn"],
    cmap="Blues"
)
plt.title("Confusion Matrix - Logistic Regression")
plt.tight_layout()
plt.show()

# Display ROC curve
RocCurveDisplay.from_predictions(y_test, y_prob)
plt.title("ROC Curve - Logistic Regression")
plt.tight_layout()
plt.show()
```

```python
# STEP 16: Evaluate Logistic Regression using 5-fold cross-validation

from sklearn.model_selection import StratifiedKFold, cross_validate
import pandas as pd

# Create 5 stratified folds
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Select evaluation metrics
scoring = {
    "accuracy": "accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "roc_auc": "roc_auc"
}

# Evaluate the complete pipeline on the training data
cv_results = cross_validate(
    logistic_model,
    X_train,
    y_train,
    cv=cv,
    scoring=scoring,
    n_jobs=-1
)

# Summarize the results across folds
cv_summary = pd.DataFrame({
    "Mean Score": [
        cv_results["test_accuracy"].mean(),
        cv_results["test_precision"].mean(),
        cv_results["test_recall"].mean(),
        cv_results["test_f1"].mean(),
        cv_results["test_roc_auc"].mean()
    ],
    "Standard Deviation": [
        cv_results["test_accuracy"].std(),
        cv_results["test_precision"].std(),
        cv_results["test_recall"].std(),
        cv_results["test_f1"].std(),
        cv_results["test_roc_auc"].std()
    ]
}, index=["Accuracy", "Precision", "Recall", "F1-Score", "ROC-AUC"])

print("5-Fold Cross-Validation Results:")
display(cv_summary.round(4))
```

```python
# STEP 17: Detect potential outliers using Z-score analysis

import numpy as np
import pandas as pd
from scipy.stats import zscore

# Select numerical columns with more than two unique values
numeric_cols = [
    col for col in df_clean.select_dtypes(include="number").columns
    if df_clean[col].nunique() > 2
]

# Calculate Z-scores for numerical features
z_scores = df_clean[numeric_cols].apply(
    lambda col: zscore(col, nan_policy="omit")
)

# Identify values with an absolute Z-score greater than 3
z_threshold = 3
outlier_mask = z_scores.abs() > z_threshold

# Count potential outliers in each feature
zscore_summary = pd.DataFrame({
    "Mean": df_clean[numeric_cols].mean(),
    "Standard Deviation": df_clean[numeric_cols].std(),
    "Potential Outliers": outlier_mask.sum()
})

print("Z-score Outlier Analysis:")
display(zscore_summary.round(3))

# Display the number of rows containing at least one potential outlier
print(
    "\nRows with at least one potential outlier:",
    int(outlier_mask.any(axis=1).sum())
)
```

```python
# STEP 18: Feature Engineering
# Create new features from existing customer data

df_engineered = df_clean.copy()

# Feature 1: Group customers by their tenure
df_engineered["TenureGroup"] = pd.cut(
    df_engineered["tenure"],
    bins=[-1, 12, 24, 48, float("inf")],
    labels=["0-12 months", "13-24 months", "25-48 months", "49+ months"]
)

# Feature 2: Count the services each customer uses
service_columns = [
    "PhoneService",
    "MultipleLines",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

df_engineered["NumServices"] = (
    df_engineered[service_columns]
    .eq("Yes")
    .sum(axis=1)
)

# Display the new features
print("Original dataset shape:", df_clean.shape)
print("Engineered dataset shape:", df_engineered.shape)

display(
    df_engineered[
        ["tenure", "TenureGroup", "NumServices", "Churn"]
    ].head(10)
)
```

```python
# STEP 19: Feature Selection using L1-Regularized Logistic Regression

from sklearn.feature_selection import SelectFromModel
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# Prepare the dataset using our engineered features
df_feature_model = df_engineered.drop(
    columns=["customerID"], errors="ignore"
).copy()

X_feature = df_feature_model.drop(columns=["Churn"])
y_feature = df_feature_model["Churn"].map({"No": 0, "Yes": 1})

# Split the data into training and testing sets
X_train_fs, X_test_fs, y_train_fs, y_test_fs = train_test_split(
    X_feature,
    y_feature,
    test_size=0.20,
    random_state=42,
    stratify=y_feature
)

# Identify numerical and categorical features
numeric_features_fs = X_train_fs.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_features_fs = X_train_fs.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

# Preprocess numerical and categorical data
numeric_transformer_fs = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer_fs = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor_fs = ColumnTransformer(transformers=[
    ("num", numeric_transformer_fs, numeric_features_fs),
    ("cat", categorical_transformer_fs, categorical_features_fs)
])

# Select features using L1 regularization
feature_selector = SelectFromModel(
    LogisticRegression(
        penalty="l1",
        solver="liblinear",
        C=0.1,
        max_iter=2000,
        random_state=42
    )
)

# Build and train the feature-selection pipeline
selection_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor_fs),
    ("selector", feature_selector)
])

selection_pipeline.fit(X_train_fs, y_train_fs)

# Display feature-selection results
feature_names = (
    selection_pipeline.named_steps["preprocessor"]
    .get_feature_names_out()
)

selected_mask = (
    selection_pipeline.named_steps["selector"]
    .get_support()
)

selected_features = feature_names[selected_mask]

print("Total processed features:", len(feature_names))
print("Selected features:", len(selected_features))
print("\nSelected features:")
for feature in selected_features:
    print(feature)
```

```python
# STEP 20: Logistic Regression Coefficient Estimation

from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
import pandas as pd
import numpy as np

# Build a Logistic Regression model using the preprocessing pipeline
logistic_model_final = Pipeline(steps=[
    ("preprocessor", preprocessor_fs),
    ("classifier", LogisticRegression(
        max_iter=2000,
        random_state=42
    ))
])

# Train the model using the training data
logistic_model_final.fit(X_train_fs, y_train_fs)

# Get feature names after preprocessing
feature_names = (
    logistic_model_final
    .named_steps["preprocessor"]
    .get_feature_names_out()
)

# Extract estimated Logistic Regression coefficients
coefficients = (
    logistic_model_final
    .named_steps["classifier"]
    .coef_[0]
)

# Create a coefficient summary
coefficient_summary = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": coefficients,
    "Odds Ratio": np.exp(coefficients)
})

# Display the most positive and most negative coefficients
coefficient_summary["Absolute Coefficient"] = (
    coefficient_summary["Coefficient"].abs()
)

coefficient_summary = coefficient_summary.sort_values(
    "Absolute Coefficient",
    ascending=False
)

print("Logistic Regression Coefficient Analysis:")
display(coefficient_summary.drop(
    columns=["Absolute Coefficient"]
).head(15).round(4))
```

```python
# STEP 21: Predict Customer Churn using Logistic Regression

# Predict churn classes for the test dataset
y_pred_final = logistic_model_final.predict(X_test_fs)

# Predict the probability of churn
y_prob_final = logistic_model_final.predict_proba(X_test_fs)[:, 1]

# Display actual and predicted results
prediction_results = pd.DataFrame({
    "Actual Churn": y_test_fs.map({0: "No", 1: "Yes"}).values,
    "Predicted Churn": pd.Series(y_pred_final).map(
        {0: "No", 1: "Yes"}
    ).values,
    "Churn Probability (%)": (y_prob_final * 100).round(2)
})

print("Customer Churn Predictions:")
display(prediction_results.head(10))
```

```python
# STEP 22: Evaluate the Final Logistic Regression Model

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt
import pandas as pd

# Calculate evaluation metrics
final_metrics = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-Score",
        "ROC-AUC"
    ],
    "Score": [
        accuracy_score(y_test_fs, y_pred_final),
        precision_score(y_test_fs, y_pred_final, zero_division=0),
        recall_score(y_test_fs, y_pred_final, zero_division=0),
        f1_score(y_test_fs, y_pred_final, zero_division=0),
        roc_auc_score(y_test_fs, y_prob_final)
    ]
})

print("Final Logistic Regression Evaluation:")
display(final_metrics.round(4))

# Display detailed classification results
print("\nClassification Report:")
print(classification_report(
    y_test_fs,
    y_pred_final,
    target_names=["No Churn", "Churn"],
    zero_division=0
))

# Plot the confusion matrix
ConfusionMatrixDisplay.from_predictions(
    y_test_fs,
    y_pred_final,
    display_labels=["No Churn", "Churn"],
    cmap="Blues"
)

plt.title("Final Logistic Regression - Confusion Matrix")
plt.tight_layout()
plt.show()
```
